```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Step 1: Files load - dhyan do .txt bhi likha hai
expr = pd.read_csv("GLDS-43_array_normalized_expression_probeset_GLmicroarray.csv")
meta = pd.read_csv("s_OSD-43.csv.txt", sep="\t")  # .txt laga diya

print("Data load ho gaya!")

# Baaki code same rahega
earth_cols = [c for c in expr.columns if "Ecol_K-12" in c and "MG" not in c]
mg_cols = [c for c in expr.columns if "MG" in c]

print("Earth samples:", len(earth_cols))
print("MG samples:", len(mg_cols))

expr["Earth_mean"] = expr[earth_cols].mean(axis=1)
expr["MG_mean"] = expr[mg_cols].mean(axis=1)
expr["log2FC"] = np.log2(expr["MG_mean"] / expr["Earth_mean"])

pvals = []
for i, row in expr.iterrows():
    t, p = stats.ttest_ind(row[mg_cols], row[earth_cols], nan_policy='omit')
    pvals.append(p)
expr["p_value"] = pvals
expr["-log10p"] = -np.log10(expr["p_value"])

print("\nTop 5 UP genes:")
print(expr[expr["log2FC"] > 1].sort_values("log2FC", ascending=False)[["SYMBOL", "log2FC"]].head())

print("\nTop 5 DOWN genes:")
print(expr[expr["log2FC"] < -1].sort_values("log2FC")[["SYMBOL", "log2FC"]].head())

plt.figure(figsize=(8,6))
plt.scatter(expr["log2FC"], expr["-log10p"], alpha=0.5, s=10)
plt.axvline(x=1, color='r', linestyle='--')
plt.axvline(x=-1, color='r', linestyle='--')
plt.axhline(y=1.3, color='r', linestyle='--') 
plt.xlabel("Log2 Fold Change")
plt.ylabel("-log10 p-value")
plt.title("Volcano Plot - E.coli Microgravity vs Earth")
plt.show()

expr.to_csv("results_log2FC.csv", index=False)
print("\nHo gaya Aiman! 🚀")
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

df = pd.read_csv('GLDS-43_array_normalized_expression_probeset_GLmicroarray.csv')
print("Data loaded. Shape:", df.shape)
```

```python
# 1. Earth aur MG columns dhundo
cols = df.columns
earth_cols = [c for c in cols if 'Earth' in c]
mg_cols = [c for c in cols if 'Microgravity' in c or 'MG' in c]
print("Earth:", earth_cols)
print("Microgravity:", mg_cols)

# 2. Numbers me convert karo
for c in earth_cols + mg_cols:
    df[c] = pd.to_numeric(df[c], errors='coerce')

# 3. log2FC aur p-value nikalo
df['Earth_mean'] = df[earth_cols].mean(axis=1)
df['MG_mean'] = df[mg_cols].mean(axis=1)
df['log2FC'] = np.log2(df['MG_mean'] / df['Earth_mean'])

from scipy import stats
tstats, pvals = stats.ttest_ind(df[mg_cols].values, df[earth_cols].values, axis=1, equal_var=False, nan_policy='omit')
df['p_value'] = pvals

# 4. Top 10 significant genes
top_genes = df.sort_values('p_value').head(10)
print("\nTop 10 genes:")
print(top_genes[['ProbeSetID', 'GeneSymbol', 'GeneTitle', 'log2FC', 'p_value']])

# 5. Save kar lo
top_genes.to_csv('top_10_genes.csv', index=False)
print("\nFile saved: top_10_genes.csv")
```
